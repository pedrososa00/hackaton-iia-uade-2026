```python
import pandas as pd

# 1. Carga directa de la ficha M14 desde el repositorio de la materia
url = "https://raw.githubusercontent.com/dharioprodigio/hackaton-iia-uade-2026/main/datos/M14.csv"
df = pd.read_csv(url)

# Separación de los conjuntos según la estructura oficial
ent = df[df.conjunto == "entrenamiento"]  # Casos 1 a 14
pru = df[df.conjunto == "prueba"]        # Casos 15 a 20


# 2. Definición de la función de decisión con los nombres exactos de las columnas
def predecir_escalamiento(row):
  intentos = row['Intentos fallidos (respuestas que no resolvieron)']
  tono = row['Tono del cliente (−1 (furioso) a 1 (cordial))']
  monto = row['Monto en disputa (USD)']

  # Regla 1: Monto alto y múltiples intentos fallidos
  if monto >= 1500 and intentos >= 3:
    return "SI"

  # Regla 2: Intentos fallidos altos con tono neutral o bajo
  elif intentos >= 5 and tono <= 0.5:
    return "SI"

  # Regla 3: Tono extremadamente furioso combinado con persistencia
  elif tono <= -0.75 and intentos >= 3:
    return "SI"

  # Caso contrario: Resuelve el agente de forma autónoma
  else:
    return "NO"


# 3. Aplicar el modelo a los datos de prueba
pru = pru.copy()
pru["prediccion"] = pru.apply(predecir_escalamiento, axis=1)

# Mostrar los resultados para verificar contra el resultado real
print(pru[['caso', 'resultado_real', 'prediccion']])
```
